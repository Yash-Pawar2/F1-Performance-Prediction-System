```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\raj\F1-Performance-Prediction-System\Data Set\Processed\F1_Proccced_Data.csv")
```

# Feature Engineering

### race phase

```python
def race_phase(lap):
    if lap <= 15:
        return "Early"
    elif lap <= 40:
        return "Middle"
    else:
        return "Late"

df["RacePhase"] = df["LapNumber"].apply(race_phase)
```

### Is Soft

```python
df["IsSoft"] = (df["Compound"] == "SOFT").astype(int)
```

### tyre age

```python
def tyre_age(x):
    if x <= 10:
        return "Fresh"
    elif x <= 25:
        return "Medium"
    else:
        return "Old"

df["TyreAge"] = df["TyreLife"].apply(tyre_age)
```

```python
df["PitStop"] = (
    df.groupby(["Season", "Race", "Driver"])["Stint"]
      .diff()
      .fillna(0)
      .ne(0)
      .astype(int)
)
```

### driver avg

```python
driver_avg = (
    df.groupby("Driver")["LapTime"]
      .mean()
      .to_dict()
)

df["DriverAvgLap"] = df["Driver"].map(driver_avg)
```

### Average Team Pace

```python
team_avg = (
    df.groupby("Team")["LapTime"]
      .mean()
      .to_dict()
)

df["TeamAvgLap"] = df["Team"].map(team_avg)
```

### Position Group

```python
def position_group(position):
    if position <= 3:
        return "Podium"
    elif position <= 10:
        return "Points"
    else:
        return "Backfield"

df["PositionGroup"] = df["Position"].apply(position_group)
```

```python
df["PositionGroup"] = pd.cut(
    df["Position"],
    bins=[0, 3, 10, 22],
    labels=["Podium", "Points", "Backfield"]
)
```

### Track conditions

```python
def track_condition(status):
    status = str(status)

    if "4" in status:
        return "Safety Car"
    elif "2" in status:
        return "Yellow Flag"
    elif "1" in status:
        return "Green Flag"
    else:
        return "Other"

df["TrackCondition"] = df["TrackStatus"].apply(track_condition)
```

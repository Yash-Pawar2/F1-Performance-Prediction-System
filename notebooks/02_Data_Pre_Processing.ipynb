```python
import pandas as pd
import glob

files = glob.glob("yearly_data/F1_*.csv")

df = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)

df.to_csv("F1_2018_2025_All.csv", index=False)

print(df.shape)
```

```python
print(df.info())
print(df.isnull().sum())
```

```python
df = df.dropna(subset=["LapTime"])
#Removing Data with no Laptimes

df = df[df["Round"] > 0]
#Remove Pre-Session Data

df["Compound"] = df["Compound"].fillna("UNKNOWN")

# Fill TyreLife with median
df["TyreLife"] = df["TyreLife"].fillna(df["TyreLife"].median())

# Fill Stint with median
df["Stint"] = df["Stint"].fillna(df["Stint"].median())

# Fill TrackStatus with mode
df["TrackStatus"] = df["TrackStatus"].fillna(df["TrackStatus"].mode()[0])

df = df.dropna(subset=["Sector1Time", "Sector2Time", "Sector3Time"])
```

```python
print(df.info())
print(df.isnull().sum())
```

```python
df.to_csv("F1_Proccced_Data.csv", index=False)
```
